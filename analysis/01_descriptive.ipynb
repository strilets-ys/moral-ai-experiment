## Feedback Summary

```python
# General feedback
data %>%
  filter(!is.na(debrief_general_feedback) & debrief_general_feedback != "" & debrief_general_feedback != "None" 
         & debrief_general_feedback != "No" & debrief_general_feedback != "NO" & debrief_general_feedback != "none" & debrief_general_feedback != "N/a" ) %>%
  dplyr::select(participant_id, condition, debrief_general_feedback)

# Persuasion descriptions (from those who noticed)
data %>%
  filter(!is.na(debrief_persuasion_description) & debrief_persuasion_description != "" & debrief_persuasion_description != "None" 
         & debrief_persuasion_description != "No" & debrief_persuasion_description != "NO" & debrief_persuasion_description != "none" & debrief_persuasion_description != "N/a") %>%
  dplyr::select(participant_id, condition, debrief_persuasion_description)

# Change descriptions (from those who changed mind)
data %>%
  filter(!is.na(debrief_change_description) & debrief_change_description != "" & debrief_change_description != "None" 
         & debrief_change_description != "No" & debrief_change_description != "NO" & debrief_change_description != "none" & debrief_change_description != "N/a") %>%
  dplyr::select(participant_id, condition, debrief_change_description)
```

## Persuasion Awareness & Self-Report

```python
# Noticed persuasion - overall
data %>%
  count(debrief_noticed_persuasion) %>%
  mutate(pct = round(n / sum(n) * 100, 1))

# Noticed persuasion - by condition
data %>%
  count(condition, debrief_noticed_persuasion) %>%
  group_by(condition) %>%
  mutate(pct = round(n / sum(n) * 100, 1)) %>%
  filter(debrief_noticed_persuasion == TRUE) %>%
  dplyr::select(condition, n_noticed = n, pct_noticed = pct)

# Self-reported mind change - overall
data %>%
  count(debrief_changed_mind) %>%
  mutate(pct = round(n / sum(n) * 100, 1))

# Self-reported mind change - by condition
data %>%
  count(condition, debrief_changed_mind) %>%
  group_by(condition) %>%
  mutate(pct = round(n / sum(n) * 100, 1)) %>%
  filter(debrief_changed_mind == TRUE) %>%
  dplyr::select(condition, n_changed = n, pct_changed = pct)
```

## Conversation Metrics

```python
# Get all chat turn count columns
turn_cols <- names(data)[grepl("^chat_turn_count_", names(data))]

# Average turns per conversation (across all dilemmas)
data %>%
  dplyr::select(all_of(turn_cols)) %>%
  pivot_longer(everything(), names_to = "dilemma", values_to = "turns") %>%
  filter(!is.na(turns) & turns > 0) %>%
  summarise(
    n_conversations = n(),
    avg_turns_per_convo = mean(turns),
    min_turns = min(turns),
    max_turns = max(turns)
  )
```

## Time Spent on Experiment

```python
# Calculate experiment duration (adjust column names if needed)
data %>%
  mutate(
    start_time = as.POSIXct(created_at),
    end_time = as.POSIXct(completed_at),
    duration_mins = as.numeric(difftime(end_time, start_time, units = "mins"))
  ) %>%
  summarise(
    avg_duration = mean(duration_mins, na.rm = TRUE),
    sd_duration = sd(duration_mins, na.rm = TRUE),
    min_duration = min(duration_mins, na.rm = TRUE),
    max_duration = max(duration_mins, na.rm = TRUE)
  )
```

## AI Trust/Usage

```python
# AI Trust (S-TIAS) summary
data %>%
  summarise(
    trust_mean = mean(debrief_stias_average, na.rm = TRUE),
    trust_sd = sd(debrief_stias_average, na.rm = TRUE),
    trust_min = min(debrief_stias_average, na.rm = TRUE),
    trust_max = max(debrief_stias_average, na.rm = TRUE)
  )

# AI usage frequency
data %>% count(debrief_ai_usage_frequency) %>% mutate(pct = round(n / sum(n) * 100, 1))

# AI tools used (split by comma, count each tool separately)
data %>%
  dplyr::select(participant_id, debrief_ai_tools_used) %>%
  filter(!is.na(debrief_ai_tools_used)) %>%
  separate_rows(debrief_ai_tools_used, sep = ",\\s*") %>%
  mutate(tool = str_trim(str_to_lower(debrief_ai_tools_used))) %>%
  count(tool, sort = TRUE) %>%
  mutate(pct = round(n / n_distinct(data$participant_id) * 100, 1))

# LLM provider distribution (which AI was used in experiment)
data %>% count(llm_provider) %>% mutate(pct = round(n / sum(n) * 100, 1))
```

## Demographics

```python
# Demographics summary
message("=== PARTICIPANT DEMOGRAPHICS ===\n")

# Age
message("Age:")
data %>%
summarise(
  n = sum(!is.na(demographics_age)),
  mean = round(mean(demographics_age, na.rm = TRUE), 1),
  sd = round(sd(demographics_age, na.rm = TRUE), 1),
  min = min(demographics_age, na.rm = TRUE),
  max = max(demographics_age, na.rm = TRUE)
) %>%
print()

# Gender
message("\nGender:")
data %>%
count(demographics_gender) %>%
mutate(pct = round(n / sum(n) * 100, 1)) %>%
print()

# Education
message("\nEducation:")
data %>%
count(demographics_education) %>%
mutate(pct = round(n / sum(n) * 100, 1)) %>%
arrange(desc(n)) %>%
print()

# Native English
message("\nNative English Speaker:")
data %>%
count(demographics_native_english) %>%
mutate(pct = round(n / sum(n) * 100, 1)) %>%
print()

# Condition distribution
message("\nCondition:")
data %>%
count(condition) %>%
mutate(pct = round(n / sum(n) * 100, 1)) %>%
print()

# LLM provider distribution
message("\nLLM Provider:")
data %>%
count(llm_provider) %>%
mutate(pct = round(n / sum(n) * 100, 1)) %>%
print()

# AI usage frequency (if available)
if ("ai_usage_frequency" %in% names(data)) {
message("\nAI Usage Frequency:")
data %>%
  count(ai_usage_frequency) %>%
  mutate(pct = round(n / sum(n) * 100, 1)) %>%
  print()
}

# Demographics visualization
demo_age <- ggplot(data, aes(x = demographics_age)) +
geom_histogram(binwidth = 5, fill = "#3498DB", color = "white") +
labs(title = "Age Distribution", x = "Age", y = "Count") +
theme_minimal()

demo_gender <- data %>%
count(demographics_gender) %>%
ggplot(aes(x = reorder(demographics_gender, -n), y = n, fill = demographics_gender)) +
geom_col() +
geom_text(aes(label = n), vjust = -0.3) +
labs(title = "Gender", x = "", y = "Count") +
theme_minimal() +
theme(legend.position = "none")

demo_education <- data %>%
count(demographics_education) %>%
ggplot(aes(x = reorder(demographics_education, n), y = n)) +
geom_col(fill = "#3498DB") +
geom_text(aes(label = n), hjust = -0.3) +
coord_flip() +
labs(title = "Education", x = "", y = "Count") +
theme_minimal()


```

```python
# =============================================================================
# Descriptive Statistics - Summary Tables
# =============================================================================

message("=== CREATING DESCRIPTIVE SUMMARY TABLES ===\n")

library(gridExtra)
library(grid)

# Calculate all stats
n_total <- nrow(data)
age_mean <- round(mean(data$demographics_age, na.rm = TRUE), 1)
age_sd <- round(sd(data$demographics_age, na.rm = TRUE), 1)
age_range <- sprintf("%d-%d", min(data$demographics_age, na.rm = TRUE), max(data$demographics_age, na.rm = TRUE))

# Gender
gender_counts <- table(data$demographics_gender)
n_female <- ifelse("female" %in% names(gender_counts), as.numeric(gender_counts["female"]), 0)
n_male <- ifelse("male" %in% names(gender_counts), as.numeric(gender_counts["male"]), 0)

# Education (using actual values from data)
edu_counts <- table(data$demographics_education)
n_hs <- ifelse("high_school" %in% names(edu_counts), as.numeric(edu_counts["high_school"]), 0)
n_some_college <- ifelse("some_college" %in% names(edu_counts), as.numeric(edu_counts["some_college"]), 0)
n_bach <- ifelse("bachelor" %in% names(edu_counts), as.numeric(edu_counts["bachelor"]), 0)
n_mast <- ifelse("master" %in% names(edu_counts), as.numeric(edu_counts["master"]), 0)
n_doc <- ifelse("doctorate" %in% names(edu_counts), as.numeric(edu_counts["doctorate"]), 0)

trust_mean <- round(mean(data$debrief_stias_average, na.rm = TRUE), 2)
trust_sd <- round(sd(data$debrief_stias_average, na.rm = TRUE), 2)

condition_counts <- table(data$condition)
llm_counts <- table(data$llm_provider)

# -----------------------------------------------------------------------------
# TABLE 1: Sample Demographics
# -----------------------------------------------------------------------------

demo_table <- data.frame(
  Variable = c("Total N", 
               "Age, M (SD)", 
               "Age range",
               "Gender: Female",
               "Gender: Male",
               "Education: High School",
               "Education: Some College",
               "Education: Bachelor's",
               "Education: Master's",
               "Education: Doctorate",
               "AI Trust (S-TIAS), M (SD)"),
  Value = c(
    as.character(n_total),
    sprintf("%.1f (%.1f)", age_mean, age_sd),
    age_range,
    sprintf("%d (%.0f%%)", n_female, n_female/n_total*100),
    sprintf("%d (%.0f%%)", n_male, n_male/n_total*100),
    sprintf("%d (%.0f%%)", n_hs, n_hs/n_total*100),
    sprintf("%d (%.0f%%)", n_some_college, n_some_college/n_total*100),
    sprintf("%d (%.0f%%)", n_bach, n_bach/n_total*100),
    sprintf("%d (%.0f%%)", n_mast, n_mast/n_total*100),
    sprintf("%d (%.0f%%)", n_doc, n_doc/n_total*100),
    sprintf("%.2f (%.2f)", trust_mean, trust_sd)
  )
)

# Create table with styling
table_theme <- ttheme_default(
  core = list(
    fg_params = list(fontsize = 14, hjust = 0, x = 0.05),
    bg_params = list(fill = c("grey95", "white"))
  ),
  colhead = list(
    fg_params = list(fontface = "bold", fontsize = 16),
    bg_params = list(fill = "grey75")
  )
)

t1 <- tableGrob(demo_table, rows = NULL, theme = table_theme)
title1 <- textGrob("Sample Demographics", gp = gpar(fontsize = 18, fontface = "bold"))

png("figures/desc_table_demographics.png", width = 1200, height = 900, res = 150)
grid.newpage()
grid.arrange(title1, t1, nrow = 2, heights = c(0.1, 0.9))
dev.off()

message("Saved: figures/desc_table_demographics.png")

# -----------------------------------------------------------------------------
# TABLE 2: Study Design / Conditions
# -----------------------------------------------------------------------------

design_table <- data.frame(
  Variable = c("Condition: Neutral",
               "Condition: Persuade",
               "Condition: Persuade (Demo)",
               "Condition: Persuade (Info)",
               "LLM: Claude (Anthropic)",
               "LLM: Qwen"),
  Value = c(
    sprintf("%d (%.0f%%)", condition_counts["neutral"], condition_counts["neutral"]/n_total*100),
    sprintf("%d (%.0f%%)", condition_counts["persuade"], condition_counts["persuade"]/n_total*100),
    sprintf("%d (%.0f%%)", condition_counts["persuade_demo"], condition_counts["persuade_demo"]/n_total*100),
    sprintf("%d (%.0f%%)", condition_counts["persuade_info"], condition_counts["persuade_info"]/n_total*100),
    sprintf("%d (%.0f%%)", llm_counts["anthropic"], llm_counts["anthropic"]/n_total*100),
    sprintf("%d (%.0f%%)", llm_counts["qwen"], llm_counts["qwen"]/n_total*100)
  )
)

t2 <- tableGrob(design_table, rows = NULL, theme = table_theme)
title2 <- textGrob("Study Design", gp = gpar(fontsize = 18, fontface = "bold"))

png("figures/desc_table_design.png", width = 1200, height = 500, res = 150)
grid.newpage()
grid.arrange(title2, t2, nrow = 2, heights = c(0.15, 0.85))
dev.off()

message("Saved: figures/desc_table_design.png")

# Print tables
message("\n=== DEMOGRAPHICS ===")
print(demo_table)
message("\n=== STUDY DESIGN ===")
print(design_table)
```
